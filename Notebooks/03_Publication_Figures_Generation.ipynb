## Figure plan
Generate publication-ready PNG and vector PDF figures from Appendix E aggregate frequencies. Figures use neutral accessible styling, explicit N, and descriptive labels. They are descriptive and do not combine distinct or conflicting questionnaire-note claims.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
OUT=Path('analysis_outputs'); OUT.mkdir(exist_ok=True)
teacher_data=[
 ('Essential values',[('Global citizenship',14),('Empathy',3),('Respect',2),('Honesty',1)]),
 ('Preferred techniques',[('Debates',9),('Storytelling',6),('Role-playing',3),('Group work',2)]),
 ('Reported obstacles',[('Student resistance',10),('Time constraints',6),('Lack of training',4)]),
 ('Curriculum guidance',[('Insufficient',16),('Sufficient',4)]),
]
```

## Teacher summary panels
Bars show counts with within-item percentages. The value item is a single summarized choice distribution as supplied; do not interpret as a multi-select denominator unless confirmed.

```python
fig,axes=plt.subplots(2,2,figsize=(11,7.2),constrained_layout=True)
colors=['#245A73','#45818E','#79A7A1','#B3C9B5']
for ax,(title,vals) in zip(axes.flat,teacher_data):
    labels=[v[0] for v in vals]; counts=[v[1] for v in vals]; n=sum(counts)
    bars=ax.barh(labels,counts,color=colors[:len(vals)],edgecolor='white')
    ax.invert_yaxis(); ax.set_title(title,loc='left',fontweight='bold'); ax.set_xlabel('Frequency (appendix N=20)')
    ax.set_xlim(0,20); ax.grid(axis='x',alpha=.2); ax.set_axisbelow(True)
    for bar,x in zip(bars,counts): ax.text(x+.25,bar.get_y()+bar.get_height()/2,f'{x} ({x/n:.0%})',va='center',fontsize=9)
fig.suptitle('Teacher questionnaire summary (Appendix E)',fontsize=15,fontweight='bold')
for ext in ('png','pdf'): fig.savefig(OUT/f'figure_teacher_summary.{ext}',dpi=300,bbox_inches='tight')
plt.show()
```

## Student experience panels
Student-table denominators are inferred from the reported row frequencies. The plots state this explicitly; validate against original records before treating as observed sample N.

```python
student_data=[
 ('Story recall',[('Recalled',17),('Did not recall',3)]),
 ('Cultural perspective',[('English helps',15),('Not sure',4),('No',1)]),
 ('Story preference',[('Clear moral',12),('Learner decides',5),('Both',3)]),
]
fig,axes=plt.subplots(1,3,figsize=(12,4.2),constrained_layout=True)
for ax,(title,vals) in zip(axes,student_data):
    labels=[v[0] for v in vals]; counts=[v[1] for v in vals]; n=sum(counts)
    bars=ax.barh(labels,counts,color=colors[:len(vals)],edgecolor='white')
    ax.invert_yaxis(); ax.set_title(title,loc='left',fontweight='bold'); ax.set_xlabel('Frequency (table total=20; inferred)')
    ax.set_xlim(0,20); ax.grid(axis='x',alpha=.2); ax.set_axisbelow(True)
    for bar,x in zip(bars,counts): ax.text(x+.2,bar.get_y()+bar.get_height()/2,f'{x} ({x/n:.0%})',va='center',fontsize=9)
fig.suptitle('Student questionnaire summary (Appendix F)',fontsize=14,fontweight='bold')
for ext in ('png','pdf'): fig.savefig(OUT/f'figure_student_summary.{ext}',dpi=300,bbox_inches='tight')
plt.show()
```

```python
expected=[OUT/f'figure_{group}_summary.{ext}' for group in ('teacher','student') for ext in ('png','pdf')]
for p in expected: assert p.exists() and p.stat().st_size>1000, f'Missing/empty: {p}'
print('Verified figure artifacts:')
for p in expected: print(p, p.stat().st_size, 'bytes')
```
