## Inter-coder agreement utility (uncomputed)
Cohen’s κ requires two independent coders labeling the same sampling units with mutually exclusive categories. The agreed Appendix procedure mentions 20% random subsample and κ≥.80, but coder-level data were not supplied. Do not report a κ until those inputs exist.

```python
def cohens_kappa(rater_a, rater_b):
    if len(rater_a)!=len(rater_b) or not rater_a: raise ValueError('Need equal, nonempty paired labels')
    labels=set(rater_a)|set(rater_b); n=len(rater_a)
    observed=sum(a==b for a,b in zip(rater_a,rater_b))/n
    expected=sum((rater_a.count(k)/n)*(rater_b.count(k)/n) for k in labels)
    return (observed-expected)/(1-expected) if expected<1 else (1.0 if observed==1 else float('nan'))
# Function unit test only—these fabricated labels are not project observations.
assert abs(cohens_kappa(['a','a','b','b'],['a','a','b','b'])-1)<1e-12
print('Agreement function smoke test passed. Project κ: NOT ESTIMABLE (no double-coded records).')
```

## Recommended audit sequence
1. Freeze and version the de-identified corpus; record exclusions and missingness.
2. Randomly select the prespecified 20% sample with a fixed seed; both coders code independently.
3. Define coding unit, category rules, multi-coding policy, and treatment of missing/ambiguous text before κ.
4. Calculate κ on the independent labels, report n and category distribution, adjudicate disagreements, revise codebook, and retain decision log.
5. Analyze remaining corpus after codebook stabilization; actively search for negative cases and link themes to research questions.
6. Report only evidence that is traceable to a verified source segment; maintain consent and confidentiality controls.
